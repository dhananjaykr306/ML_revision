**DATA PREPROCESSING**

**ENCODING**

```python
categorical_columns = df.select_dtypes(include=['object']).columns
continous_columns = df.select_dtypes(include=['int64','float64']).columns
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

for col in categorical_columns:
    df[col] = label_encoder.fit_transform(df[col])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig,ax = plt.subplots(1,3,figsize=(10,5))
sns.histplot(df['Annual Income (k$)'],kde=True,bins=50,ax=ax[0])
ax[0].set_title('Annual Income (k$)')
sns.histplot(df['Age'],kde=True,bins=50,ax=ax[1])
ax[1].set_title('Age')
sns.histplot(df['Spending Score (1-100)'],kde=True,bins=50,ax=ax[2])
ax[2].set_title('Spending Score (1-100)')
plt.show()
```

# Scaling 

```python
num_columns = [col for col in df.columns if df[col].dtype != object]
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
```

```python
df[num_columns] = scaler.fit_transform(df[num_columns])
```

**MODEL**

```python
from sklearn.cluster import KMeans
df = df[["Annual Income (k$)","Spending Score (1-100)"]]
inertia=[]

for i in range(1,11):
  model = KMeans(n_clusters=i,random_state=42)
  model.fit(df)
  inertia.append(model.inertia_)
```

```python
import matplotlib.pyplot as plt

plt.plot(range(1,11),inertia)
plt.show()
```

FROM THE ELBOW METHOD WE CAN SAY THAT ,K=5 WORKS BETTER

```python
model=KMeans(n_clusters=5,random_state=42)
model.fit(df)
```

ADDING CLUTER COLUMN TO THE DATAFRAME

```python
df['Cluster']=model.labels_
```

GET THE CENTROIDS

```python
centroids = model.cluster_centers_
centroids
```

# Visualizing the clusters (2D Annual_Income Spending Score,)

```python
import matplotlib.pyplot as plt
import seaborn as sns


fig = plt.figure(figsize=(10, 10,))

sns.scatterplot(x=df['Annual Income (k$)'], y=df['Spending Score (1-100)'], data=df,hue=df['Cluster'], palette='viridis')
sns.scatterplot(x=centroids[:, 0],y= centroids[:, 1],  c='red', marker='*', s=200,label='Cluster Centers')


plt.show()
```
