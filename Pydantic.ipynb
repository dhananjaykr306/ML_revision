Pydantic is a Python library used for data validation and settings management, with an emphasis on type safety. It helps developers enforce data schemas by validating and parsing Python data structures against specified types, making it especially useful for APIs, data pipelines, and configurations.

### Key Features:
1. **Type Validation**: Ensures input data matches the defined schema.
2. **Data Parsing**: Converts data into the desired types (e.g., parsing strings into dates or integers).
3. **Custom Validation**: Allows defining custom validation rules for specific fields.
4. **Settings Management**: Can load configurations from environment variables or `.env` files.
5. **Serialization**: Supports serialization and deserialization of data.


### Validation Features
1. **Default Values**: Fields can have default values.
2. **Optional Fields**: Use `Optional` or `None` for optional fields.
3. **Custom Validators**: Use `@validator` to define custom rules.


### Use Cases:
1. API request/response validation.
2. Database schema validation.
3. Environment configuration parsing.

```python
from pydantic import BaseModel,EmailStr,validator
```

```python
class User(BaseModel):
    name : str
    email : EmailStr
    account_id : int

    @validator("account_id")
    def validate_account_id(cls,value):
        if value <=0:
            raise ValueError(f"account_id must be positive{value}")
        return value

user = User(name="Dhananjay",email="dhananjaykr306@gmail.com",account_id=1234)
print(user)
```

```python
# pydantic model to json
```

```python
user_json_str = user.json()
print(user_json_str)
```

```python
# pydantic model to dictionary
```

```python
user_json_model = user.dict()
print(user_json_model)
```

```python
# json string to pydantic model

json_str = '{"name":"Dhananjay","email":"dhananjaykr306@gmail.com","account_id":1234}'
user = user.parse_raw(json_str)
print(user)
```
