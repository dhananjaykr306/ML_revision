## Baye's Theorem

Bayes' Theorem is a mathematical formula used to calculate conditional probabilities. It describes how to update the probability of a hypothesis based on new evidence. The theorem is expressed as:

P(A|B) = {P(B|A)*P(A)} / P(B)

P(B) = {P(B|A)* P(A)} + {P(B|~ A)*P(~ A)}

Where:
- \(P(A|B)\): Probability of event \(A\) given \(B\).
- \(P(B|A)\): Probability of \(B\) given \(A\).
- \(P(A)\): Prior probability of \(A\).
- \(P(B)\): Probability of \(B\).
- ( P(B|~ A) \) is the probability of \( B \) given (~ A \)
- \( P(\~ A) \) is the probability of (~ A ).

# 1. Diagnostic Test Accuracy

Question:

A diagnostic test for a disease has:

Sensitivity (true positive rate): 95%

Specificity (true negative rate): 90%

Disease prevalence in the population: 1%

If a person tests positive, what is the probability they actually have the disease?

```python
P_disease = 0.01  # Prevalence
P_no_disease = 1 - P_disease
P_positive_given_disease = 0.95  # Sensitivity
P_positive_given_no_disease = 1 - 0.90  # 1 - Specificity

# Total probability of testing positive (P(B))
P_positive = (P_positive_given_disease * P_disease) + (P_positive_given_no_disease * P_no_disease)

# Bayes' theorem: P(A|B)
P_disease_given_positive = (P_positive_given_disease * P_disease) / P_positive

print(f"Probability of having the disease if tested positive: {P_disease_given_positive:.2%}")
```

# 2. Spam Email Classification

Question:

A spam filter determines that:

- 70% of emails are non-spam.
- The probability of a spam email containing the word "lottery" is 0.5.
- The probability of a non-spam email containing the word "lottery" is 0.01.

If an email contains the word "lottery," what is the probability it is spam?

```python
# Given data
P_spam = 0.3
P_non_spam = 0.7
P_lottery_given_spam = 0.5
P_lottery_given_non_spam = 0.01

# Calculate P(lottery)
P_lottery = (P_lottery_given_spam * P_spam + P_lottery_given_non_spam * P_non_spam)

# Bayes' theorem
P_spam_given_lottery = ((P_lottery_given_spam * P_spam) / P_lottery)

print(f"Probability the email is spam if it contains 'lottery': {P_spam_given_lottery:}")
```

# 3 Weather Prediction

Question

In a city, it rains 20% of the time, and it doesn’t rain 80% of the time. A weather app claims that it gives a 90% probability of "rain predicted" when it rains, and only a 10% probability of "rain predicted" when it doesn't rain. If the weather app predicts rain, what is the probability that it will actually rain?

```python
# Given data
P_rain = 0.20  # Probability of rain
P_no_rain = 1 - P_rain  # Probability of no rain
P_predict_rain_given_rain = 0.90  # Sensitivity
P_predict_rain_given_no_rain = 0.10  # False positive rate

# Total probability of predicting rain (P(B))
P_predict_rain = (P_predict_rain_given_rain * P_rain) + (P_predict_rain_given_no_rain * P_no_rain)

# Bayes' theorem: P(A|B)
P_rain_given_predict_rain = (P_predict_rain_given_rain * P_rain) / P_predict_rain

print(f"Probability of rain given the app predicts rain: {P_rain_given_predict_rain:.2%}")
```

# 4. Manufacturing Defects

Question:

A factory produces 5% defective items. A quality control test detects defective items 98% of the time but falsely labels 2% of non-defective items as defective. If an item fails the test, what is the probability it is defective?

```python
# Given data
P_defective = 0.05
P_non_defective = 1 - P_defective
P_fail_given_defective = 0.98
P_fail_given_non_defective = 0.02

# Bayes' theorem
P_defective_given_fail = ((P_fail_given_defective * P_defective) /((P_fail_given_defective * P_defective) + (P_fail_given_non_defective * P_non_defective)))

print(f"Probability an item is defective if it fails the test: {P_defective_given_fail:.2%}")
```

# 5. COVID-19 Testing

Question:

A COVID-19 test has 99% sensitivity and 95% specificity. If the infection rate in the population is 2%, what is the probability that a person who tests positive actually has COVID-19?

```python
# Given data
P_covid = 0.02
P_no_covid = 1 - P_covid
P_positive_given_covid = 0.99  # Sensitivity
P_positive_given_no_covid = 1 - 0.95  

# Bayes' theorem
P_covid_given_positive = ((P_positive_given_covid * P_covid) / ((P_positive_given_covid * P_covid) + (P_positive_given_no_covid * P_no_covid)))

print(f"Probability of having COVID-19 if tested positive: {P_covid_given_positive:.2%}")
```

Question

In a neighbourhood, 90% children were falling sick due flu and 10% due to measles and no other disease. The probability of observing rashes for measles is 0.95 and for flu is 0.08. If a child develops rashes, find the child’s probability of having flu.

```python
P_Flu = 0.90
P_M = 0.10
P_R_given_flu = 0.08
P_R_given_Measles = 0.10

# Bayes' theorem
P_R = (P_R_given_flu * P_Flu + P_R_given_Measles*P_M)

P_F_given_Rashes = (P_R_given_flu*P_Flu)/(P_R)
print(f"Probability of having Flu given that child having Rashes : {P_F_given_Rashes :.2f}")
```
