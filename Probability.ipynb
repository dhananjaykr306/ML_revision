 1. ##### A group of 60 students is randomly split into 3 classes of equal size. All partitions are equally likely. Jack and Jill are two students belonging to that group.

##### What is the probability that Jack and Jill will end up in the same class ?


```python
from math import comb

# Total number of students and the class size
total_students = 60
class_size = 20

# Probability that Jack and Jill will end up in the same class
numerator = comb(total_students - 2, class_size - 2)  
denominator = comb(total_students, class_size)  

probability = numerator / denominator
print(f"Probability that Jack and Jill will end up in the same class: {probability * 100:.2f}%")
```

2. ##### Jack has two coins in his hand. Out of the two coins, one is a real coin and the second one is a faulty one with Tails on both sides. He blindfolds himself to choose a random coin and tosses it in the air. The coin falls down with Tails facing upwards.

##### What is the probability that this tail is shown by the faulty coin ?


```python
# Given probabilities
# Probability of choosing the faulty coin
P_A = 1 / 2  
# Probability of showing Tails given the faulty coin
P_B_given_A = 1
# Probability of showing Tails given the real coin
P_B_given_real = 1 / 2 
# Probability of choosing the real coin
P_real_coin = 1 / 2

# Total probability of showing Tails
P_B = P_A * P_B_given_A + P_real_coin * P_B_given_real

# Conditional probability: P(A|B) = (P(B|A) * P(A)) / P(B)
P_A_given_B = (P_B_given_A * P_A) / P_B

# Print the result as a percentage
print(f"Probability that the tail is shown by the faulty coin: {P_A_given_B * 100:.2f}%")
```

3. ##### Rob has fever and the doctor suspects it to be typhoid. To be sure, the doctor wants to conduct the test. The test results are positive when the patient actually has typhoid 80% of the time. The test gives positive when the patient does not have typhoid 10% of the time.

##### If 1% of the population has typhoid, what is the probability that Rob has typhoid provided he tested positive?


```python
# Given probabilities
# Probability that Rob has typhoid
P_A = 0.01  
# Probability of testing positive given that Rob has typhoid
P_B_given_A = 0.80  
 # Probability of testing positive given that Rob does not have typhoid
P_B_given_not_A = 0.10 
# Probability that Rob does not have typhoid
P_not_A = 1 - P_A  

# Total probability of testing positive
P_B = P_B_given_A * P_A + P_B_given_not_A * P_not_A

# Conditional probability: P(A|B) = (P(B|A) * P(A)) / P(B)
P_A_given_B = (P_B_given_A * P_A) / P_B

# Print the result as a percentage
print(f"Probability that Rob has typhoid given he tested positive: {P_A_given_B * 100:.2f}%")
```

4. ##### Hospital records show that 75% of patients suffering from a disease die due to that disease. What is the probability that 4 out of the 6 randomly selected patients recover?

```python
from math import comb

# number of patients
n = 6  
# number of recoveries
k = 4  
# probability of recovery
p_recovery = 0.25  

# binomial probability formula: P(X = k) = comb(n, k) * p^k * (1-p)^(n-k)
probability = comb(n, k) * (p_recovery ** k) * ((1 - p_recovery) ** (n - k))
print(f"Probability that exactly 4 out of 6 patients recover: {probability:.4f}")
```

6. ##### In a class of 30 students, approximately what is the probability that two of the students have their birthday on the same day (defined by the same day and month) (assuming it’s not a leap year)?


```python
from math import comb,factorial

# Total number of ways to assign birthdays to 30 students
total_outcomes = 365 ** 30

# Counting favorable outcomes
choose_2_students = comb(30, 2)
choose_28_days = comb(364, 28)
assign_28_birthdays = factorial(28)

# Number of favorable outcomes
favorable_outcomes = 365 * choose_2_students * choose_28_days * assign_28_birthdays

# Probability calculation
probability = favorable_outcomes / total_outcomes

print(f"Probability that exactly two students share a birthday: {probability:.10f}")
```

7. ##### A coin of diameter 1-inches is thrown on a table covered with a grid of lines each two inches apart. What is the probability that the coin lands inside a square without touching any of the lines of the grid?

8. ##### Suppose you were interviewed for a technical role. 50% of the people who sat for the first interview received the call for the second interview. 95% of the people who got a call for a second interview felt good about their first interview. 75% of people who did not receive a second call, also felt good about their first interview.

If you felt good after your first interview, what is the probability that you will receive a second interview call?


```python
def probability_of_second_call():
    # Given probabilities
    P_A = 0.50  # Probability of receiving a second cal
    P_B_given_A = 0.95  # Probability of feeling good given second call
    P_B_given_Ac = 0.75  # Probability of feeling good given no second call
    P_Ac = 1 - P_A  # Probability of not receiving a second call

    # Total probability of feeling good
    P_B = (P_B_given_A * P_A) + (P_B_given_Ac * P_Ac)

    # Bayes' Theorem: P(A | B)
    P_A_given_B = (P_B_given_A * P_A) / P_B

    return P_A_given_B

prob = probability_of_second_call()
print(f"The probability of receiving a second call given that you feel good is: {prob:.4f}")
```

9. ##### Some test scores follow a normal distribution with a mean of 18 and a standard deviation of 6. What proportion of test takers have scored between 18 and 24?

```python
from scipy.stats import norm

# Parameters
mu = 18  # Mean
sigma = 6  # Standard deviation

# Scores
x1 = 18
x2 = 24

# Calculate z-scores
z1 = (x1 - mu) / sigma
z2 = (x2 - mu) / sigma

# Calculate cumulative probabilities
p1 = norm.cdf(z1)  # P(Z <= z1)
p2 = norm.cdf(z2)  # P(Z <= z2)

# Proportion between 18 and 24
proportion = p2 - p1
print(f"The proportion of test takers who scored between 18 and 24 is: {proportion:.4f}")
```

10. ##### Suppose a life insurance company sells a $240,000 one year term life insurance policy to a 25-year old female for $210. The probability that the female survives the year is .999592.

Find the expected value of this policy for the insurance company.

```python
def expected_value():
    # Given values
    premium = 210  # Premium collected
    payout = 240000  # Payout if the customer dies
    prob_survive = 0.999592  # Probability of survival
    prob_die = 1 - prob_survive  # Probability of death

    # Profit calculations
    profit_survive = premium  # Profit if the customer survives
    profit_die = premium - payout  # Profit if the customer dies

    # Expected value
    expected_value = (profit_survive * prob_survive) + (profit_die * prob_die)
    return expected_value

# Calculate and print the result
ev = expected_value()
print(f"The expected value of this policy for the insurance company is: ${ev:.2f}")
```
